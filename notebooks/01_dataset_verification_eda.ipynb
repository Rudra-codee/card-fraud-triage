## 2. In-Depth Data Quality & Anomaly Audit

In this section, we verify:
1. Infinite values (positive or negative) across numeric features.
2. Duplicate row counts using both `keep='first'` (standard) and `keep=False` (all rows in duplicate clusters).
3. Duplicate breakdown grouped by the `Class` target.
4. Non-negative constraints for `Time` and `Amount`.
5. Prevalence of zero-amount transactions.
6. Constant or zero-variance columns.
7. Target validity (ensuring only binary classes {0, 1} exist).

```python
# 1. Infinite values check
numeric_cols = df.select_dtypes(include=[np.number]).columns
pos_inf = np.isposinf(df[numeric_cols]).sum().sum()
neg_inf = np.isneginf(df[numeric_cols]).sum().sum()
print(f"Positive infinite values: {pos_inf}")
print(f"Negative infinite values: {neg_inf}")

# 2. Duplicate rows audit
dup_default = df.duplicated().sum()
dup_all = df.duplicated(keep=False).sum()
print(f"\nDuplicate row copies (df.duplicated().sum()): {dup_default}")
print(f"Total rows involved in duplicate clusters (df.duplicated(keep=False).sum()): {dup_all}")

# 3. Duplicate breakdown by Class
dup_clusters = df[df.duplicated(keep=False)]
print("\nDuplicate cluster rows breakdown by Class:")
print(dup_clusters['Class'].value_counts())

fraud_dups_all = dup_clusters[dup_clusters['Class'] == 1]
fraud_dups_copies = df[df.duplicated() & (df['Class'] == 1)]
print(f"\nTotal fraud rows in duplicate clusters: {len(fraud_dups_all)}")
print(f"Redundant fraud duplicate copies: {len(fraud_dups_copies)}")
print(f"Unique fraud clusters: {len(fraud_dups_all) - len(fraud_dups_copies)}")

# 4. Value boundary checks for Time and Amount
print(f"\nNegative Time values: {(df['Time'] < 0).sum()}")
print(f"Negative Amount values: {(df['Amount'] < 0).sum()}")
print(f"Zero Amount values: {(df['Amount'] == 0).sum()} ({((df['Amount'] == 0).mean()*100):.3f}%)")

# 5. Constant columns check
constant_cols = [c for c in df.columns if df[c].nunique() <= 1]
print(f"\nConstant columns (zero variance): {constant_cols if constant_cols else 'None'}")

# 6. Target validity check
unique_classes = df['Class'].unique()
print(f"\nTarget unique values: {unique_classes.tolist()}")
assert set(unique_classes).issubset({0, 1}), "Unexpected classes found!"
```

## 3. Comparative Statistical Analysis: Legitimate vs. Fraudulent Transactions

We compute and contrast descriptive statistics for `Time` and `Amount` between legitimate transactions (`Class = 0`) and fraudulent transactions (`Class = 1`).

```python
legit = df[df['Class'] == 0]
fraud = df[df['Class'] == 1]

# Summary statistics for Amount across classes
amount_comp = pd.DataFrame({
    'Metric': ['Count', 'Mean', 'Std Dev', 'Min', '25% (Q1)', '50% (Median)', '75% (Q3)', 'Max', 'IQR', 'Skewness'],
    'Legitimate (Class 0)': [
        f"{len(legit):,}",
        f"${legit['Amount'].mean():.2f}",
        f"${legit['Amount'].std():.2f}",
        f"${legit['Amount'].min():.2f}",
        f"${legit['Amount'].quantile(0.25):.2f}",
        f"${legit['Amount'].median():.2f}",
        f"${legit['Amount'].quantile(0.75):.2f}",
        f"${legit['Amount'].max():.2f}",
        f"${legit['Amount'].quantile(0.75) - legit['Amount'].quantile(0.25):.2f}",
        f"{legit['Amount'].skew():.2f}"
    ],
    'Fraud (Class 1)': [
        f"{len(fraud):,}",
        f"${fraud['Amount'].mean():.2f}",
        f"${fraud['Amount'].std():.2f}",
        f"${fraud['Amount'].min():.2f}",
        f"${fraud['Amount'].quantile(0.25):.2f}",
        f"${fraud['Amount'].median():.2f}",
        f"${fraud['Amount'].quantile(0.75):.2f}",
        f"${fraud['Amount'].max():.2f}",
        f"${fraud['Amount'].quantile(0.75) - fraud['Amount'].quantile(0.25):.2f}",
        f"{fraud['Amount'].skew():.2f}"
    ]
}).set_index('Metric')

print("=== AMOUNT COMPARISON ===")
display(amount_comp)

# Summary statistics for Time across classes
time_comp = pd.DataFrame({
    'Metric': ['Count', 'Mean (s)', 'Std Dev (s)', 'Min (s)', '25% (s)', '50% (Median s)', '75% (s)', 'Max (s)', 'Range (hours)'],
    'Legitimate (Class 0)': [
        f"{len(legit):,}",
        f"{legit['Time'].mean():.1f}",
        f"{legit['Time'].std():.1f}",
        f"{legit['Time'].min():.1f}",
        f"{legit['Time'].quantile(0.25):.1f}",
        f"{legit['Time'].median():.1f}",
        f"{legit['Time'].quantile(0.75):.1f}",
        f"{legit['Time'].max():.1f}",
        f"{(legit['Time'].max() - legit['Time'].min()) / 3600:.2f}"
    ],
    'Fraud (Class 1)': [
        f"{len(fraud):,}",
        f"{fraud['Time'].mean():.1f}",
        f"{fraud['Time'].std():.1f}",
        f"{fraud['Time'].min():.1f}",
        f"{fraud['Time'].quantile(0.25):.1f}",
        f"{fraud['Time'].median():.1f}",
        f"{fraud['Time'].quantile(0.75):.1f}",
        f"{fraud['Time'].max():.1f}",
        f"{(fraud['Time'].max() - fraud['Time'].min()) / 3600:.2f}"
    ]
}).set_index('Metric')

print("\n=== TIME COMPARISON ===")
display(time_comp)

print(f"\nZero-amount transactions in Legitimate: {(legit['Amount'] == 0).sum()} ({(legit['Amount'] == 0).mean()*100:.3f}%)")
print(f"Zero-amount transactions in Fraud: {(fraud['Amount'] == 0).sum()} ({(fraud['Amount'] == 0).mean()*100:.3f}%)")
```

## 4. Exploratory Data Visualizations

In this section, we generate clear visual representations of:
1. Target class imbalance (linear and log scale).
2. Transaction amount distribution (raw scale with log counts).
3. Transformed `log1p(Amount)` distribution.
4. Temporal distribution of transactions across the 48-hour recording window.
5. Boxplot of `Amount` and `log1p(Amount)` grouped by `Class`.
6. Temporal density distribution comparison between Legitimate and Fraudulent transactions.

```python
fig_dir = "../results/figures" if os.path.exists("../results") else "results/figures"
os.makedirs(fig_dir, exist_ok=True)

class_labels = {0: 'Legitimate (0)', 1: 'Fraud (1)'}
df['Class_Label'] = df['Class'].map(class_labels)
df['log1p_Amount'] = np.log1p(df['Amount'])

# --- Figure 1: Class Distribution ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))
counts = df['Class_Label'].value_counts()
colors = ['#2b5c8f', '#d9534f']

bars1 = ax1.bar(counts.index, counts.values, color=colors, edgecolor='black', alpha=0.85, width=0.45)
ax1.set_title('Class Distribution (Linear Scale)', fontsize=12, fontweight='bold', pad=10)
ax1.set_ylabel('Transaction Count', fontsize=11)
ax1.set_xlabel('Transaction Class', fontsize=11)
for bar in bars1:
    yval = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2.0, yval + 3500, f'{yval:,}\n({yval/len(df)*100:.3f}%)', 
             ha='center', va='bottom', fontsize=10, fontweight='bold')
ax1.set_ylim(0, 320000)

bars2 = ax2.bar(counts.index, counts.values, color=colors, edgecolor='black', alpha=0.85, width=0.45)
ax2.set_yscale('log')
ax2.set_title('Class Distribution (Log10 Scale)', fontsize=12, fontweight='bold', pad=10)
ax2.set_ylabel('Transaction Count (log10 scale)', fontsize=11)
ax2.set_xlabel('Transaction Class', fontsize=11)
for bar in bars2:
    yval = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2.0, yval * 1.35, f'{yval:,}', 
             ha='center', va='bottom', fontsize=10, fontweight='bold')
ax2.set_ylim(1, 1000000)

plt.suptitle('Figure 1: Extreme Class Imbalance in Credit Card Transactions', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
fig.savefig(os.path.join(fig_dir, '01_class_imbalance.png'), dpi=300, bbox_inches='tight')
plt.show()

# --- Figure 2: Amount Histogram ---
fig, ax = plt.subplots(figsize=(8.5, 4.5))
ax.hist(df['Amount'], bins=100, color='#2b5c8f', edgecolor='black', alpha=0.85)
ax.set_title('Figure 2: Distribution of Transaction Amount (Raw Scale, Log Counts)', fontsize=12, fontweight='bold', pad=10)
ax.set_xlabel('Transaction Amount ($)', fontsize=11)
ax.set_ylabel('Frequency (Log Scale)', fontsize=11)
ax.set_yscale('log')
ax.axvline(df['Amount'].median(), color='gold', linestyle='--', linewidth=2, label=f'Median: ${df["Amount"].median():.2f}')
ax.axvline(df['Amount'].mean(), color='red', linestyle='-', linewidth=2, label=f'Mean: ${df["Amount"].mean():.2f}')
ax.legend(frameon=True, facecolor='white', loc='upper right', fontsize=10)
plt.tight_layout()
fig.savefig(os.path.join(fig_dir, '02_amount_histogram.png'), dpi=300, bbox_inches='tight')
plt.show()

# --- Figure 3: log1p(Amount) Histogram ---
fig, ax = plt.subplots(figsize=(8.5, 4.5))
ax.hist(df['log1p_Amount'], bins=80, color='#17becf', edgecolor='black', alpha=0.85)
ax.set_title('Figure 3: Distribution of log1p(Amount)', fontsize=12, fontweight='bold', pad=10)
ax.set_xlabel('log1p(Amount) [ln(Amount + 1)]', fontsize=11)
ax.set_ylabel('Frequency', fontsize=11)
ax.axvline(df['log1p_Amount'].median(), color='darkblue', linestyle='--', linewidth=2, label=f'Median: {df["log1p_Amount"].median():.2f}')
ax.axvline(df['log1p_Amount'].mean(), color='crimson', linestyle='-', linewidth=2, label=f'Mean: {df["log1p_Amount"].mean():.2f}')
ax.legend(frameon=True, facecolor='white', loc='upper right', fontsize=10)
plt.tight_layout()
fig.savefig(os.path.join(fig_dir, '03_amount_log1p_histogram.png'), dpi=300, bbox_inches='tight')
plt.show()

# --- Figure 4: Time Histogram ---
fig, ax = plt.subplots(figsize=(9.5, 4.5))
ax.hist(df['Time'], bins=96, color='#2ca02c', edgecolor='black', alpha=0.85)
ax.set_title('Figure 4: Distribution of Transaction Time Across 48 Hours', fontsize=12, fontweight='bold', pad=10)
ax.set_xlabel('Elapsed Time from First Transaction (seconds)', fontsize=11)
ax.set_ylabel('Transaction Count per ~30-min Window', fontsize=11)
ax.axvline(86400, color='darkred', linestyle=':', linewidth=2, label='24 Hours (Day 1 End: 86,400s)')
ax.axvline(172800, color='darkred', linestyle=':', linewidth=2, label='48 Hours (Day 2 End: 172,800s)')
ax.legend(frameon=True, facecolor='white', loc='upper right', fontsize=10)
plt.tight_layout()
fig.savefig(os.path.join(fig_dir, '04_time_histogram.png'), dpi=300, bbox_inches='tight')
plt.show()

# --- Figure 5: Amount Boxplot by Class ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))
sns.boxplot(x='Class_Label', y='Amount', data=df, hue='Class_Label', palette=['#2b5c8f', '#d9534f'], legend=False, ax=ax1, flierprops={'marker': 'o', 'markersize': 2, 'alpha': 0.3})
ax1.set_title('Raw Amount ($) by Class\n(Outliers visible up to $25.6k)', fontsize=11, fontweight='bold')
ax1.set_ylabel('Amount ($)', fontsize=11)
ax1.set_xlabel('Transaction Class', fontsize=11)

sns.boxplot(x='Class_Label', y='log1p_Amount', data=df, hue='Class_Label', palette=['#2b5c8f', '#d9534f'], legend=False, ax=ax2, flierprops={'marker': 'o', 'markersize': 2, 'alpha': 0.3})
ax2.set_title('log1p(Amount) by Class\n(Lower Median for Fraud)', fontsize=11, fontweight='bold')
ax2.set_ylabel('log1p(Amount)', fontsize=11)
ax2.set_xlabel('Transaction Class', fontsize=11)

plt.suptitle('Figure 5: Transaction Amount Distribution Comparison Across Classes', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
fig.savefig(os.path.join(fig_dir, '05_amount_boxplot_by_class.png'), dpi=300, bbox_inches='tight')
plt.show()

# --- Figure 6: Temporal Density Comparison ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6.5), sharex=True)
sns.histplot(df[df['Class']==0]['Time'], bins=96, kde=True, stat='density', color='#1f77b4', ax=ax1, alpha=0.6)
ax1.set_title('Legitimate Transactions (Class 0): Pronounced Diurnal Cycle with Nighttime Lulls', fontsize=11, fontweight='bold')
ax1.set_ylabel('Density', fontsize=10)
ax1.axvline(86400, color='darkred', linestyle=':', label='24h Boundary')
ax1.legend(loc='upper right')

sns.histplot(df[df['Class']==1]['Time'], bins=96, kde=True, stat='density', color='#d62728', ax=ax2, alpha=0.6)
ax2.set_title('Fraudulent Transactions (Class 1): Activity Continues Throughout Night Hours', fontsize=11, fontweight='bold')
ax2.set_ylabel('Density', fontsize=10)
ax2.set_xlabel('Elapsed Time (seconds)', fontsize=11)
ax2.axvline(86400, color='darkred', linestyle=':', label='24h Boundary')
ax2.legend(loc='upper right')

plt.suptitle('Figure 6: Temporal Density Comparison Across Classes', fontsize=13, fontweight='bold', y=0.99)
plt.tight_layout()
fig.savefig(os.path.join(fig_dir, '06_time_density_comparison.png'), dpi=300, bbox_inches='tight')
plt.show()
```

## 5. Audit Findings & Collaborative Decisions Required

### Summary of Key Metrics
* **Total Transactions:** 284,807
* **Features:** 30 numerical predictors (`Time`, `V1`–`V28`, `Amount`) + 1 binary target (`Class`).
* **Missing / Infinite Values:** Exactly 0 across all features.
* **Class Imbalance:** 284,315 legitimate (99.827%) vs. 492 fraudulent (0.173%). Ratio: ~578:1.
* **Duplicates:** 1,081 duplicate copies (`keep='first'`), 1,854 total rows involved across clusters (`keep=False`).
  * Class 0 duplicates: 1,822 rows (1,062 redundant copies).
  * Class 1 duplicates: 32 rows (19 redundant copies across 13 distinct transaction profiles).

### Decision Needed: Duplicate Rows Handling
* **Observed Fact:** 19 duplicate fraud entries exist in the dataset.
* **Trade-off:**
  * If dropped: We lose 19 of our 492 fraud samples (3.86% of all positive examples).
  * If kept: If identical rows fall into both train and test sets during random splitting, it causes data leakage and optimistic test evaluation.
  * Recommendation: Do not drop duplicates immediately. Perform stratified train/test split at the *unique cluster* level or discuss with the team whether these represent bot attack bursts or pipeline duplicate telemetry.
